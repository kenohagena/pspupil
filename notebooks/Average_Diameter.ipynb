```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import scipy.stats as stats
```

```python
# FOOOF included, no z-scoring
edition = '/Volumes/psp_data/PSP/Pupil_Preprocessed_2024-06-17/'
```

```python
cov = pd.read_hdf(join(edition, 'PowerFrequencies', 'CoV.hdf'), key='data')
cov = cov.loc[(cov.session == 'Baseline')&(cov.group.isin(['PSP', 'PD', 'Control']))&(cov.condition == 'bp_non_normalized')]

cov = cov.groupby(['group', 'subject']).mean().reset_index() #average per subject across runs
```

```python
average_diam = cov.groupby('group').mean().round(1)
average_diam
```

```python
std_diam = cov.groupby('group').std()
std_diam
```

```python
average_diam.loc['Control', 'mean'].round(1)
```

```python
# ANOVA mit scipy
f_stat, p_val = stats.f_oneway(cov.loc[cov.group== 'PSP', 'mean'].values,
                               cov.loc[cov.group== 'PD', 'mean'].values,
                               cov.loc[cov.group== 'Control', 'mean'].values)

print("F-Wert:", f_stat, "p-Wert:", p_val)
```

```python
'The mean pupil diameter did not significantly differ across groups (PSP {0} ± {3} pixel, PD {1} ±{4} pixel, HC {2} ±{5} pixel; F={6}, p={7})'.format(average_diam.loc['PSP', 'mean'].round(1),
                                                                                                                                        average_diam.loc['PD', 'mean'].round(1),
                                                                                                                                        average_diam.loc['Control', 'mean'].round(1),
                                                                                                                                        std_diam.loc['PSP', 'mean'].round(1),
                                                                                                                                       std_diam.loc['PD', 'mean'].round(1),
                                                                                                                                       std_diam.loc['Control', 'mean'].round(1),
                                                                                                                                                     f_stat.round(2),
                                                                                                                                                     p_val.round(2))
```
