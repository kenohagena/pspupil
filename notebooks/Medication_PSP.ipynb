```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
from scipy.stats import ttest_rel
```

```python
data = pd.read_csv('/Users/kenohagena/Documents/Forschung/PSP/data/meta/Medikation_PSPRS_Marc_KH.csv', delimiter=";")
data = data.iloc[:-1, :-3]
```

```python
data['PSP-RS Difference'] = data.loc[:, 'PSP-RS ON'] - data.loc[:, 'PSP-RS OFF']
data['PSP-RS Difference %'] = data.loc[:, 'PSP-RS Difference'] / data.loc[:, 'PSP-RS OFF']
```

### Medication in PSP patients

```python
'Out of {0} tested PSP patients {1} took dopaminergic medication, {2} took amantadin, {3} took both,{4} took none'.format(data.count()[0], 
          data.loc[data.dopaminerg == 'Ja'].count()[0],
          data.loc[data.amantadin == 'Ja'].count()[0],
          data.loc[(data.amantadin == 'Ja')&(data.dopaminerg == 'Ja')].count()[0],
          data.loc[(data.amantadin == 'Nein')&(data.dopaminerg == 'Nein')].count()[0])
```

```python
'LEDD on median {0} ({1} | {2})'.format(data.LEDD.median(), data.LEDD.quantile(0.25), data.LEDD.quantile(0.75))
```

### Repeated measures T Test

```python
t_test = ttest_rel(data.loc[:, 'PSP-RS ON'].values, data.loc[:, 'PSP-RS OFF'].values)
```

```python
'Difference between ON and OFF in PSP-RS was on median {0} ({1} | {2})'.format(data.loc[:, 'PSP-RS Difference'].median(), 
                                                                               data.loc[:, 'PSP-RS Difference'].quantile(0.75), 
                                                                               data.loc[:, 'PSP-RS Difference'].quantile(0.25))
```

```python
'% Difference between ON and OFF in PSP-RS was on median {0}% ({1} | {2})'.format(data.loc[:, 'PSP-RS Difference %'].median().round(2)*100, 
                                                                               data.loc[:, 'PSP-RS Difference %'].quantile(0.75).round(5)*100, 
                                                                               data.loc[:, 'PSP-RS Difference %'].quantile(0.25).round(5)*100)
```

```python
'Repeated measures T-test revealed however that this difference was signidficant: T={0}, p={1}'.format(t_test[0], t_test[1])
```
